### Business Problem Statement

#### 1.Which region generates the highest sales?

```python
region_sales = df_cleaned.groupby('Region')['Sales'].sum().reset_index()
region_sales = region_sales.sort_values(by ='Sales', ascending=False)
```

```python
plt.figure(figsize=(13,6))
ax = sns.barplot(data=region_sales, x='Region', y='Sales',palette='rocket',width=0.3)

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')
    
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.show()
```

### ~Observation

#### West region generated the highest total sales, followed by East, Central, and South. South recorded the lowest total sales among all regions.

### ~Business Insight

#### The West region is the strongest revenue contributor. Understanding the strategies, customer preferences, and product demand in this region can help replicate its success in lower-performing regions.

### ~Recommendation

#### The company should continue investing in the West region while investigating why South is underperforming. Marketing campaigns, inventory planning, and sales strategies from the West could be adapted for the South region.

#### Q6.How do sales change month by month?

```python
monthly_sales = df_cleaned.groupby('Month')['Sales'].sum().reset_index()
```

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]
```

```python
monthly_sales["Month"] = pd.Categorical(
    monthly_sales["Month"],
    categories=month_order,
    ordered=True
)

monthly_sales = monthly_sales.sort_values("Month")
```

```python
plt.figure(figsize=(13,6))
sns.lineplot(data=monthly_sales, x='Month', y='Sales',color='black',marker='o',  linewidth=2.5)


plt.title('Monthly Total Sales')
plt.xlabel('Month')
plt.ylabel('Total_Sales')
```

### ~Observation

#### Sales were lowest in February, increased significantly in March, remained relatively stable from April to August, and then showed a strong upward trend from September onwards. November recorded the highest monthly sales, followed by December.

### ~Business Insight

#### The business experiences stronger sales during the last quarter (Q4), especially in November and December. This suggests the influence of seasonal demand, year-end shopping, or promotional campaigns that drive higher customer purchases.

### ~Recommendation

#### The company should prepare additional inventory, marketing campaigns, and operational resources before the high-demand months of November and December. It should also investigate the reasons for the low sales in February and identify strategies to improve performance during slower periods.

### 9.Which product category is most preferred for each shipping mode?

```python
pref_ship_category = df_cleaned.groupby(['Ship Mode','Category'])['Order ID'].count().reset_index(name='Order_Count')
```

```python
plt.figure(figsize=(13,6))
ax = sns.barplot(data=pref_ship_category, x='Category', y='Order_Count',hue='Ship Mode',palette='rocket')

for container in ax.containers:
    ax.bar_label(container, fmt='%.0f')

plt.title('Most Preferred Product Category by Shipping Mode')
plt.xlabel('Product Category')
plt.ylabel('Order Count')
plt.show()
```

### ~Observation

#### Standard Class is the most preferred shipping mode across all three product categories. Office Supplies received the highest number of orders through Standard Class, followed by Furniture and Technology. Same Day shipping is the least preferred option for every product category.

### ~Business Insight

#### Customers generally prefer cost-effective shipping methods over premium delivery options. Standard Class dominates regardless of product category, indicating that customers prioritize economical shipping even when purchasing Technology products.

### ~Recommendation

#### The company should continue optimizing Standard Class shipping by ensuring timely delivery and operational efficiency. To increase adoption of faster shipping options, the business can introduce promotional discounts or special offers for First Class and Same Day delivery.
