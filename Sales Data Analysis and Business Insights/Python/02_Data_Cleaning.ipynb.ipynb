## Removing Postle code NULL values

## Date Time Conversions
### Oder_Date and Ship_Date

```python
df_clean['Order Date'] = pd.to_datetime(df_clean['Order Date'])
```

## Challenge Faced
### Why it happens:
### Date columns often mix ambiguous formats like DD-MM-YYYY and MM-DD-YYYY. Pandas assumes one format (like %m-%d-%Y) for the whole column. It works fine until it hits a value like 15-04-2018, where 15 can't be a month — causing a ValueError.

### How i solved it:

### Inspected the raw date values instead of guessing the format.
### Found that day consistently comes before month (DD-MM-YYYY).
### Parsed the column explicitly using that format:

```python
df_clean['Order Date'].sample(30).tolist()
```

```python
df_clean['Order Date'] = pd.to_datetime(df_clean['Order Date'], format='%d-%m-%Y')
```

```python
df_clean['Order Date'].isna().sum()
```

```python
df_clean['Ship Date'].sample(30).tolist()
```

```python
df_clean['Ship Date'] = pd.to_datetime(df_clean['Ship Date'], format='%d-%m-%Y')
```

```python
df_clean['Ship Date'].isna().sum()
```

### After Solving Problem

```python
df_clean['Order Date'] = pd.to_datetime(df_clean['Order Date'])
df_clean['Ship Date'] = pd.to_datetime(df_clean['Ship Date'])
```
